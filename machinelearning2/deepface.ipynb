```python
import cv2
import matplotlib.pyplot as plt
from pathlib import Path
from deepface import DeepFace
```

```python
# 1. Defina a pasta (troque pelo caminho da sua pasta)
pasta = Path(r"/home/lucas/Imagens/Webcam")

# 2. Pega todos os arquivos da pasta
arquivos = [f for f in pasta.iterdir() if f.is_file()]

if arquivos:
    # 3. Encontra o arquivo com a maior (mais recente) data de modificação
    arquivo_mais_recente = max(arquivos, key=lambda f: f.stat().st_mtime)
   
    # 4. Pega o path completo como texto (string)
    caminho_completo = str(arquivo_mais_recente)


# Carrega a imagem
img_path = caminho_completo
img = cv2.imread(img_path)

# Analisa a foto
resultados = DeepFace.analyze(
    img_path=img_path,
    actions=['age', 'gender', 'emotion', 'race'],
    enforce_detection=False,
)

# Desenha para cada rosto detectado
for rosto in resultados:
    idade = rosto["age"]
    genero = rosto["dominant_gender"]
    emocao = rosto["dominant_emotion"]
    raca = rosto["dominant_race"]
    regiao = rosto["region"]

    x, y, w, h = regiao["x"], regiao["y"], regiao["w"], regiao["h"]

    # Retângulo ao redor do rosto
    cv2.rectangle(img, (x, y), (x + w, y + h), (0, 255, 0), 2)

    # Cada linha de texto precisa de uma chamada separada
    linhas = [
        f"Age: {idade}",
        f"Gender: {genero}",
        f"Emotion: {emocao}",
        f"Race: {raca}"
    ]

    for i, linha in enumerate(linhas):
        pos_y = y - 10 - (len(linhas) - 1 - i) * 25  # empilha as linhas acima do rosto
        cv2.putText(
            img, linha, (x, pos_y),
            cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 255, 0), 2
        )

cv2.imshow("Resultado", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
